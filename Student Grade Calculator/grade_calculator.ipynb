**Write load_students(filepath) returning a list of dicts. Handle FileNotFoundError.**

```python
import csv

def load_students(filepath):
    students = []

    try:
        with open(filepath, "r") as file:
            reader = csv.DictReader(file)

            for row in reader:
                student = {
                    "Name": row["Name"],
                    "Math": float(row["Math"]),
                    "Science": float(row["Science"]),
                    "English": float(row["English"]),
                    "History": float(row["History"]),
                    "PE": float(row["PE"])
                }

                students.append(student)

    except FileNotFoundError:
        print(f"Error: File '{filepath}' was not found.")


    return students
```

**Write calculate_stats(scores) returning: total, average, highest, lowest, grade.**

```python
def calculate_stats(scores):

    total = sum(scores)
    average = sum(scores) / len(scores)
    highest = max(scores)
    lowest = min(scores)

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    elif average >= 50:
        grade = "E"
    else:
        grade = "F"

    return total, average, highest, lowest, grade
    
```

**Write display_report(students) printing a formatted table: Name, Average, Grade, Pass/Fail.**

**Write save_report(students, output_path) saving the table + class summary to .txt.**

```python
def get_student_result(student):
    """Calculate statistics for one student."""

    scores = [
        student["Math"],
        student["Science"],
        student["English"],
        student["History"],
        student["PE"]
    ]

    total, average, highest, lowest, grade = calculate_stats(scores)

    return {
        "Name": student["Name"],
        "Total": total,
        "Average": average,
        "Highest": highest,
        "Lowest": lowest,
        "Grade": grade,
        "Result": "Pass" if average >= 50 else "Fail"
    }
```

```python
def display_report(students):
    ### Display the student grade report.

    print()
    print("=" * 60)
    print("STUDENT GRADE REPORT")
    print("=" * 60)

    print(f"{'Name':<15}{'Average':<12}{'Grade':<10}{'Result':<10}")
    print("-" * 60)

    for student in students:
        result = get_student_result(student)

        print(
            f"{result['Name']:<15}"
            f"{result['Average']:<12.2f}"
            f"{result['Grade']:<10}"
            f"{result['Result']:<10}"
        )

    print("=" * 60)
```

```python
def save_report(students, output_path="grades_report.txt"):
    
    results = [get_student_result(student) for student in students]

    if not results:
        print("No student data available.")
        return

    total_students = len(results)
    passed_students = sum(
        1 for result in results if result["Result"] == "Pass"
    )
    failed_students = total_students - passed_students

    class_average = sum(
        result["Average"] for result in results
    ) / total_students

    with open(output_path, "w", encoding="utf-8") as file:
        file.write("=" * 60 + "\n")
        file.write("STUDENT GRADE REPORT\n")
        file.write("=" * 60 + "\n\n")

        file.write(
            f"{'Name':<15}{'Average':<12}"
            f"{'Grade':<10}{'Result':<10}\n"
        )

        file.write("-" * 60 + "\n")

        for result in results:
            file.write(
                f"{result['Name']:<15}"
                f"{result['Average']:<12.2f}"
                f"{result['Grade']:<10}"
                f"{result['Result']:<10}\n"
            )

        file.write("\n" + "=" * 60 + "\n")
        file.write("CLASS SUMMARY\n")
        file.write("=" * 60 + "\n")

        file.write(f"Total Students : {total_students}\n")
        file.write(f"Passed         : {passed_students}\n")
        file.write(f"Failed         : {failed_students}\n")
        file.write(f"Class Average  : {class_average:.2f}\n")

    print(f"\nReport saved successfully to '{output_path}'.")
```

**Add main() orchestrating all steps.**

```python
def main():
    ## Run the student grade calculator.

    print("Student Grade Calculator")

    filepath = input(
        "Enter input CSV file path "
        "(press Enter for 'students_input.csv'): "
    ).strip()

    if not filepath:
        filepath = "students_input.csv"

    students = load_students(filepath)

    if not students:
        print("No student records found.")
        return

    display_report(students)
    save_report(students)
```

```python
if __name__ == "__main__":
    main()
```
